```python
N = int(input())
inorder = list(map(int,input().split()))
postorder = list(map(int,input().split()))

preorder = []#정답 순회

mainroot = postorder[-1]#최상위 루트
preorder.append(mainroot)

lefttree = [] #왼쪽 트리
righttree = [] #오른쪽 트리
cnt = (N-1)//2 #서브트리의 노드 개수


for i in range(len(postorder)):
  if postorder[i] == mainroot:
    break
  if i < cnt:
      lefttree.append(postorder[i])
  else:
      righttree.append(postorder[i])

preorder.append(lefttree.pop()) # 서브트리 루트 저장
if len(lefttree) != 0: #서브트리 노드 개수가 1이 아니면
  for i in lefttree:
    preorder.append(i) #나머지 노드 저장

preorder.append(righttree.pop())
if len(righttree) != 0:
  for i in righttree:
    preorder.append(i)

print(*preorder)
```

```python
from collections import deque

def DFS(lines,V,visited):
  visited.append(V)
  for i in range(len(lines)):
    if lines[i][0] == V:
      if lines[i][1] not in visited:
        DFS(lines,lines[i][1],visited)
    if lines[i][1] == V:
      if lines[i][0] not in visited:
        DFS(lines,lines[i][0],visited)

def BFS(lines, V, visited):
    queue = deque([V])  # 큐를 사용하여 탐색할 노드를 저장
    visited.append(V)    # 시작 노드를 방문 목록에 추가

    while queue:
        node = queue.popleft()  # 큐에서 노드를 꺼냄
        for i in range(len(lines)):
            if lines[i][0] == node and lines[i][1] not in visited:
                visited.append(lines[i][1])  # 방문하지 않은 노드를 방문 목록에 추가
                queue.append(lines[i][1])    # 큐에 추가하여 나중에 탐색
            elif lines[i][1] == node and lines[i][0] not in visited:
                visited.append(lines[i][0])  # 방문하지 않은 노드를 방문 목록에 추가
                queue.append(lines[i][0])    # 큐에 추가하여 나중에 탐색



N,M,V = map(int,input().split()) # 정점의 개수, 간선의 개수, 탐색 시작할 정점의 번호

visited = []

lines = [list(map(int,input().split())) for i in range(M)]

DFS(lines,V,visited)
print(*visited)

visited = []
BFS(lines,V,visited)
print(*visited)
```
